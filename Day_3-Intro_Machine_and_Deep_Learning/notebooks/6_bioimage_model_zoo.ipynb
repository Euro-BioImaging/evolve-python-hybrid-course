# The BioImage Model Zoo

The [BioImage Model Zoo](https://bioimage.io/#/) (BMZ) is a community repository of pre-trained deep learning models for bioimage analysis. 

Models are contributed by the community and can be browsed and tested directly on the website.

### BioImage.IO
 
The bioimageio provides two Python packages for working with BMZ models programmatically:
 
- [`bioimageio.spec`](https://pypi.org/project/bioimageio.spec/) — for reading and writing the BMZ model specification format
- [`bioimageio.core`](https://pypi.org/project/bioimageio.core/) — for loading and running BMZ models


In the [Bioimageio GitHub](https://github.com/bioimage-io/core-bioimage-io-python) repo you can find example notebooks that can be used locally (if required dependecies are installed) or inside a Colab notebook.

## Explore further: ZeroCostDL4Mic, DL4MicEverywhere, BiaPy

**ZeroCostDL4Mic** is a collection of GoogleColab notebooks. The notebooks are designed for researchers that have little or no coding expertise to quickly test, train and use popular Deep-Learning networks. Although Python coding knowledge is beneficial in case of issues and troubleshooting.
- [ZeroCostDL4Mic wiki](https://github.com/HenriquesLab/ZeroCostDL4Mic/wiki) 


**DL4MicEverywhere** allows conversion of the ZeroCostDL4Mic notebooks to Jupyter Notebooks using a custom script, that also packages the python environment requirements in a docker image. This, combined with notebook and release versioning, allows for easy deployment and reproducibility of workflows across space and time. 
- [DL4MicEverywhere GitHub](https://github.com/HenriquesLab/DL4MicEverywhere)


**BiaPy** is a deep learning framework focused on reproducible training and inference pipelines for bioimage analysis tasks such as segmentation, denoising, and restoration. BiaPy can be used through a GUI, but also provides ready-to-use Google Colab notebooks. BiaPy allows running of pretrained models from the BioImage Model Zoo (BMZ).
- [BiaPy GitHub](https://github.com/BiaPyX/biapy), [BiaPy page](https://biapyx.github.io/)
