# Exercise 1: Image thresholding

1. Import the relevant packages: numpy, matplotlib and skimage

````python
import numpy as np
import matplotlib.pyplot as plt
import skimage
````

2. Import the image from this url address: https://cildata.crbs.ucsd.edu/media/images/13585/13585.tif.

````python
image = skimage.io.imread('https://cildata.crbs.ucsd.edu/media/images/13585/13585.tif')
image.shape
````

3. Create a new variable called ```image_green``` that contains only the green channel by slicing/indexing the array

````python
image_green = image[:, :, 1]
````

4. Plot this image

````python
plt.imshow(image_green, cmap='grey')
````

5. Plot a histogram of the image

````python
plt.hist(image_green.flatten(), bins=50, range=(0, 256), density=True); # tip: use semicolon to suppress output
````

6. Try to determine an appropriate threshold to recover the large structures in the image and create a binary image using this threshold. Compare the binary image with the original image using `stackview.switch` (the images to be compared need to be passed to the function as a list).

````python
threshold = 50

image_green_thresholded = image_green > threshold

import stackview
stackview.switch([image_green, image_green_thresholded])
````

7. There are other algorithms for automated thresholding than the one of Otsu. Use the function `skimage.filters.try_all_threshold` to compare different thresholding methods.

````python
# nice function to try all thresholding methods:

skimage.filters.try_all_threshold(image_green, figsize=(10, 8), verbose=False);
````

8. Choose one of the thresholding methods and apply it. You can find the corresponding function in the [Scikit-image documentation](http://scikit-image.org/docs/dev/api/skimage.filters.html) (methods for determining thresholds start with ```threshold_```).

````python
threshold = skimage.filters.threshold_yen(image_green)

print('The threshold was determined as:', threshold)

image_green_thresholded = image_green > threshold

stackview.switch([image_green, image_green_thresholded])
````
